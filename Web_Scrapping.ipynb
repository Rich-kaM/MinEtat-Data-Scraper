# Import the necessary libraries

```python
import requests
from bs4 import BeautifulSoup as soup
import os
```

# Load data from website 

```python
url ="https://www.congovirtuel.com/ecole8.php"
palmares = requests.get(url)
soup = soup(palmares.text, "html.parser")
data = soup.findAll("div", attrs={"class":"article"})
data

```

# data cleanned 

```python
data = soup.findAll("a")
data
```

# finding all data finising with .pdf on my dataset

```python
pdf_links = [a['href'] for a in data if 'href' in a.attrs and a['href'].endswith('.pdf')]
print(pdf_links)
```

# downloading data and save it in local 

```python
download_data = "MinEtat-Data-Scraper/Data.pdf"
os.makedirs(download_data, exist_ok=True)
for link in data:
    href = link.get('href')
    if href and href.endswith(".pdf"):  
        file_url = href if href.startswith("http") else f"https://www.congovirtuel.com/{href}"
        file_name = os.path.join(download_data, href.split("/")[-1])

        try:
            response = requests.get(file_url, stream=True)
            response.raise_for_status()  
            with open(file_name, 'wb') as file:
                for chunk in response.iter_content(chunk_size=8192):
                    file.write(chunk)
            print(f"Downloaded: {file_name}")
        except Exception as e:
            print(f"Failed to download {file_url}: {e}")
            

            
```
